# Application: Discrete Fourier Transform Matrix Visualization

Visualize DFT matrices as complex functions to reveal their beautiful circular symmetry.

**Prerequisites**: Notebook #1 | **Time**: ~15 min

```python
import numpy as np
import matplotlib.pyplot as plt
import complexplorer as cp
```

## Mathematical Background

**DFT Matrix**: $W_N^{jk}$ where $W_N = e^{-2\pi i/N}$

- $W_N$ is the principal Nth root of unity
- Matrix is unitary: $W^* W = N \cdot I$
- Reveals structure of Fourier transform
- Connection to FFT algorithm

## DFT Matrix Generation

```python
def dft_matrix(N):
    """Generate NxN DFT matrix.
    
    Returns complex matrix where entry (j,k) = exp(-2πi jk/N)
    """
    j, k = np.meshgrid(np.arange(N), np.arange(N))
    W_N = np.exp(-2j * np.pi / N)
    return W_N ** (j * k)

# Test with N=8
W8 = dft_matrix(8)
print(f"W8 shape: {W8.shape}")
print(f"W8[0,0] = {W8[0,0]}  (should be 1)")
print(f"W8[1,1] = {W8[1,1]:.4f}  (should be e^{{-2πi/8}})")
```

## 2×4 Grid Visualization (YOUR REQUEST)

Visualize DFT matrices for N = 32, 64, 128, 256, 512, 1024, 2048, 4096

**Layout**: 2 rows × 4 columns (exactly as requested)

```python
# Sizes to visualize (USER'S SPECIFIC REQUEST)
sizes = [32, 64, 128, 256, 512, 1024, 2048, 4096]

# Colormap to reveal structure
cmap = cp.Phase(phase_sectors=12, auto_scale_r=True)

# Create 2×4 grid
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for idx, N in enumerate(sizes):
    print(f"Generating N={N}...", end=' ')
    
    # Generate DFT matrix
    W = dft_matrix(N)
    
    # Visualize as complex function
    extent = [0, N, 0, N]
    rgb = cmap.rgb(W)
    
    axes[idx].imshow(rgb, origin='lower', extent=extent, aspect='equal')
    axes[idx].set_title(f"N = {N}", fontsize=14)
    axes[idx].set_xlabel("k (frequency)", fontsize=10)
    axes[idx].set_ylabel("j (time)", fontsize=10)
    
    print("✓")

fig.suptitle("DFT Matrix Visualization: 2×4 Grid", fontsize=16)
plt.tight_layout()
plt.show()

print("\n✨ Notice the circular symmetry and periodic structure!")
```

## Circular Symmetry Exploration

Why circular? Because $W_N = e^{-2\pi i/N}$ is an Nth root of unity - it lives on the unit circle!

```python
# Compare colormaps to highlight symmetry
N = 256
W = dft_matrix(N)

colormaps = [
    ('Phase (auto-scaled)', cp.Phase(phase_sectors=12, auto_scale_r=True)),
    ('PolarChessboard', cp.PolarChessboard(phase_sectors=16, spacing=0.15)),
    ('AnalogousWedge', cp.AnalogousWedge(H_center=0.55, H_wedge=0.25)),
]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for idx, (name, cmap_test) in enumerate(colormaps):
    rgb = cmap_test.rgb(W)
    axes[idx].imshow(rgb, origin='lower', aspect='equal')
    axes[idx].set_title(f"{name}\n(N={N})", fontsize=11)
    axes[idx].axis('off')

plt.suptitle("DFT Matrix with Different Colormaps", fontsize=14)
plt.tight_layout()
plt.show()

print("\n💡 PolarChessboard reveals the radial structure beautifully!")
```

## Structure Observations

Key patterns in DFT matrices:

1. **Diagonal patterns**: $W^{jj} = W^{j^2}$ creates parabolic phase
2. **Periodicity**: Matrix repeats with period N (both axes)
3. **Symmetry**: $W^{jk} = W^{kj}$ (symmetric)
4. **Modulus**: All entries have $|W^{jk}| = 1$ (unit circle)
5. **Roots of unity**: Phase wraps N times around circle

## Extensions

### IDFT Matrix (Conjugate Transpose)

```python
N = 128
W_dft = dft_matrix(N)
W_idft = np.conj(W_dft.T) / N  # Inverse DFT

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

cmap = cp.Phase(phase_sectors=12, auto_scale_r=True)

ax1.imshow(cmap.rgb(W_dft), origin='lower', aspect='equal')
ax1.set_title("DFT Matrix", fontsize=12)
ax1.axis('off')

ax2.imshow(cmap.rgb(W_idft * N), origin='lower', aspect='equal')  # Scale for vis
ax2.set_title("IDFT Matrix (conjugate transpose)", fontsize=12)
ax2.axis('off')

plt.suptitle("DFT and Inverse DFT Matrices", fontsize=14)
plt.tight_layout()
plt.show()
```

### Other Structured Matrices

```python
N = 128

# Hadamard matrix (real-valued, ±1 entries)
from scipy.linalg import hadamard
H = hadamard(N)

# Toeplitz matrix
from scipy.linalg import toeplitz
c = np.exp(-1j * 2 * np.pi * np.arange(N) / N)
T = toeplitz(c)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
cmap = cp.Phase(phase_sectors=12, auto_scale_r=True)

axes[0].imshow(cmap.rgb(dft_matrix(N)), origin='lower', aspect='equal')
axes[0].set_title("DFT Matrix", fontsize=12)
axes[0].axis('off')

axes[1].imshow(cmap.rgb(H.astype(complex)), origin='lower', aspect='equal')
axes[1].set_title("Hadamard Matrix", fontsize=12)
axes[1].axis('off')

axes[2].imshow(cmap.rgb(T), origin='lower', aspect='equal')
axes[2].set_title("Toeplitz Matrix", fontsize=12)
axes[2].axis('off')

plt.suptitle("Structured Matrices Comparison", fontsize=14)
plt.tight_layout()
plt.show()
```
