## Setup

````python
import numpy as np
import matplotlib.pyplot as plt
import complexplorer as cp

# Check PyVista availability
try:
    import pyvista as pv
    print("✓ PyVista is available")
    HAS_PYVISTA = True
except ImportError:
    print("⚠️ PyVista not installed")
    print("   Install with: pip install 'complexplorer[pyvista]'")
    HAS_PYVISTA = False

# Standard function and colormap
def f(z):
    return (z - 1) / (z**2 + z + 1)

domain = cp.Disk(radius=2)
cmap = cp.Phase(phase_sectors=6, auto_scale_r=True)
````

## Why PyVista?

**Performance comparison**: PyVista is 15-30x faster than matplotlib for 3D rendering.

**Quality**: Better antialiasing, lighting, and interactivity.

## Inline Mode (Development)

Use `notebook=True` for inline display in Jupyter. This mode:
- ✅ Shows plots inline (convenient for development)
- ✅ Doesn't block execution
- ⚠️ Lower quality (aliasing issues)

**Best for**: Notebooks, development, quick iteration

````python
if HAS_PYVISTA:
    # Inline 3D landscape
    cp.plot_landscape_pv(
        domain, f, cmap=cmap,
        z_max=5,
        resolution=400,
        notebook=True,  # Inline mode
        show=True
    )
    print("✓ Inline plot rendered above")
else:
    print("Install PyVista to see 3D visualization")
````

## Interactive Mode (Publication Quality)

Use `notebook=False` for external windows. This mode:
- ✅ Full quality (proper antialiasing)
- ✅ Interactive (mouse rotation, zoom)
- ⚠️ Blocks execution until window closed

**Best for**: Final figures, presentations, screenshots

**Note**: We won't run this in the notebook to avoid blocking, but here's how:

```python
cp.plot_landscape_pv(
    domain, f, cmap=cmap,
    z_max=5,
    resolution=800,
    notebook=False,  # External window
    show=True
)
# Window opens - rotate with mouse, close to continue
```

## Side-by-Side Landscapes

Compare domain and codomain in 3D:

````python
if HAS_PYVISTA:
    cp.pair_plot_landscape_pv(
        domain, f, cmap=cmap,
        z_max=5,
        resolution=400,
        notebook=True,
        show=True,
        window_size=(1200, 500)
    )
    print("✓ Side-by-side 3D landscapes")
    print("   Left: Domain (identity function)")
    print("   Right: Codomain (f(z))")
````

## Riemann Sphere Introduction

Visualize functions on the extended complex plane (ℂ ∪ {∞}):

````python
if HAS_PYVISTA:
    cp.riemann_pv(
        f,
        resolution=300,
        modulus_mode='constant',  # Flat sphere (phase only)
        notebook=True,
        show=True
    )
    print("✓ Riemann sphere (constant height - phase only)")
    print("   See Notebook #6 for full Riemann sphere deep dive")
````
