# Complete Colormap Guide

Explore all 13 colormap families in Complexplorer v2.0.

**Prerequisites**: Notebook #1 | **Time**: ~25 min

```python
import numpy as np
import matplotlib.pyplot as plt
import complexplorer as cp

def f(z):
    return (z - 1) / (z**2 + z + 1)

domain = cp.Disk(radius=2)
```

## All 13 Colormaps

```python
colormaps = [
    ('Phase', cp.Phase(phase_sectors=6, auto_scale_r=True)),
    ('Chessboard', cp.Chessboard(spacing=0.3)),
    ('PolarChessboard', cp.PolarChessboard(phase_sectors=8, spacing=0.3)),
    ('LogRings', cp.LogRings(log_spacing=0.3)),
    ('PerceptualPastel', cp.PerceptualPastel(phase_sectors=6, auto_scale_r=True)),
    ('Isoluminant', cp.Isoluminant(L=0.6, C=0.15)),
    ('CubehelixPhase', cp.CubehelixPhase(phase_sectors=6, auto_scale_r=True)),
    ('AnalogousWedge', cp.AnalogousWedge(H_center=0.55, H_wedge=0.2)),
    ('DivergingWarmCool', cp.DivergingWarmCool(phase_sectors=6, auto_scale_r=True)),
    ('InkPaper', cp.InkPaper(phase_strength=0.05)),
    ('EarthTopographic', cp.EarthTopographic(phase_sectors=6, auto_scale_r=True)),
    ('FourQuadrant', cp.FourQuadrant(phase_sectors=6, auto_scale_r=True)),
]

fig, axes = plt.subplots(3, 4, figsize=(16, 12))
axes = axes.flatten()

for idx, (name, cmap) in enumerate(colormaps):
    cp.plot(domain, f, cmap=cmap, ax=axes[idx], resolution=400)
    axes[idx].set_title(name, fontsize=10)

axes[-1].set_visible(False)
fig.suptitle("All 13 Colormap Families", fontsize=16)
plt.tight_layout()
plt.show()
```

## Use Case Guide

- **Print**: PerceptualPastel, CubehelixPhase
- **Web**: Phase, EarthTopographic  
- **Scientific**: CubehelixPhase (grayscale-safe)
- **Artistic**: AnalogousWedge, InkPaper
