## Basic Imports

````python
import numpy as np
import matplotlib.pyplot as plt
import complexplorer as cp

print(f"Complexplorer version: {cp.__version__}")
print("✓ Ready to visualize complex functions!")
````

## Standard Demo Function

Throughout these notebooks, we'll use this standard function:

$$f(z) = \frac{z - 1}{z^2 + z + 1}$$

This function has:
- **One zero** at $z = 1$
- **Two poles** at $z = e^{\pm 2\pi i/3}$ (complex conjugate pair)
- Interesting phase wrapping and pole structure

````python
# Standard demo function
def f(z):
    return (z - 1) / (z**2 + z + 1)
````

## Your First Visualization

Let's create our first plot! We'll visualize the function on a rectangular domain.

````python
# Create a domain (region of complex plane to visualize)
domain = cp.Rectangle(4, 4)  # 4×4 rectangle centered at origin

# Create visualization
fig, ax = plt.subplots(figsize=(6, 6))
cp.plot(domain, f, ax=ax)
ax.set_title(r"$f(z) = \frac{z-1}{z^2+z+1}$", fontsize=14)
plt.tight_layout()
plt.show()
````

### Understanding the Color Encoding

**Phase (argument) → Hue**:
- Red = positive real (phase = 0)
- Cyan = negative real (phase = π)
- Green/Yellow = positive imaginary (phase = π/2)
- Blue/Magenta = negative imaginary (phase = -π/2)

**Key observations**:
- **Zero at z=1**: Colors converge (all phases meet)
- **Poles at z=e^{±2πi/3}**: Colors emanate (phase wraps)

## Working with Different Domains

Complexplorer supports various domain shapes:

````python
# Rectangle domain
rect = cp.Rectangle(re_length=4, im_length=3, center=0+0j)

# Disk domain
disk = cp.Disk(radius=2, center=0+0j)

# Annulus (ring) domain
annulus = cp.Annulus(inner_radius=0.5, outer_radius=2, center=0+0j)

# Visualize the same function on different domains
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

cp.plot(rect, f, ax=axes[0])
axes[0].set_title("Rectangle Domain", fontsize=12)

cp.plot(disk, f, ax=axes[1])
axes[1].set_title("Disk Domain", fontsize=12)

cp.plot(annulus, f, ax=axes[2])
axes[2].set_title("Annulus Domain", fontsize=12)

plt.tight_layout()
plt.show()
````

## Enhanced Phase Portraits

Basic phase portraits show only the argument (hue). **Enhanced phase portraits** also encode the modulus (magnitude) using radial patterns.

### New in v2.0: Auto-Scaling!

The `auto_scale_r=True` parameter automatically calculates spacing to create square cells around the unit circle.

````python
# Basic phase portrait (hue only)
basic_cmap = cp.Phase()

# Enhanced with phase sectors (angular wedges)
phase_sectors_cmap = cp.Phase(phase_sectors=6)

# Enhanced with auto-scaling (NEW in v2.0!)
auto_scaled_cmap = cp.Phase(phase_sectors=6, auto_scale_r=True)

# Compare them
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

cp.plot(disk, f, cmap=basic_cmap, ax=axes[0])
axes[0].set_title("Basic Phase Portrait\n(Hue only)", fontsize=11)

cp.plot(disk, f, cmap=phase_sectors_cmap, ax=axes[1])
axes[1].set_title("With Phase Sectors\n(6 angular wedges)", fontsize=11)

cp.plot(disk, f, cmap=auto_scaled_cmap, ax=axes[2], resolution=600)
axes[2].set_title("Auto-Scaled\n(Square cells around |z|=1)", fontsize=11)

plt.tight_layout()
plt.show()

print("\n✨ The auto-scaled version creates a checkerboard-like pattern!")
print("   Each square cell represents equal changes in phase and modulus.")
````

### Adjusting Phase Sectors

The `phase_sectors` parameter controls how many angular wedges divide the complex plane:

````python
# Try different numbers of phase sectors
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, n in enumerate([3, 6, 8, 12, 16, 24]):
    cmap = cp.Phase(phase_sectors=n, auto_scale_r=True)
    cp.plot(disk, f, cmap=cmap, ax=axes[idx], resolution=500)
    axes[idx].set_title(f"phase_sectors = {n}", fontsize=11)

plt.tight_layout()
plt.show()

print("\nRecommended values:")
print("  • phase_sectors=6: Good balance (default for examples)")
print("  • phase_sectors=12: Finer angular resolution")
print("  • phase_sectors=24: Very detailed phase structure")
````

## Side-by-Side: Domain and Codomain

Often it's helpful to see how the function transforms the domain:

````python
# Create enhanced phase colormap
cmap = cp.Phase(phase_sectors=6, auto_scale_r=True)

# Side-by-side plot
fig = cp.pair_plot(
    domain, 
    f, 
    cmap=cmap, 
    title=r"$f(z) = \frac{z-1}{z^2+z+1}$",
    resolution=600,
    figsize=(12, 5)
)
plt.tight_layout()
plt.show()

print("\nLeft: Domain (input z)")
print("Right: Codomain (output f(z))")
print("\nNotice how the grid in the domain gets distorted in the codomain!")
````

## Exploring Different Functions

Let's try some classic complex functions:

### Polynomial: $f(z) = z^4 - 1$

This function has 4 zeros (4th roots of unity):

````python
def polynomial(z):
    return z**4 - 1

cmap = cp.Phase(phase_sectors=8, auto_scale_r=True)
fig = cp.pair_plot(domain, polynomial, cmap=cmap, title=r"$f(z) = z^4 - 1$", resolution=600)
plt.tight_layout()
plt.show()

print("Notice the 4-fold symmetry around the zeros!")
````

### Transcendental: $f(z) = \sin(z)$

The sine function has infinitely many zeros along the real axis:

````python
def sine_function(z):
    return np.sin(z)

# Use wider domain to see more structure
wide_domain = cp.Rectangle(8, 6)
cmap = cp.Phase(phase_sectors=6, auto_scale_r=True)

fig = cp.pair_plot(wide_domain, sine_function, cmap=cmap, title=r"$f(z) = \sin(z)$", resolution=600)
plt.tight_layout()
plt.show()

print("Notice the periodic zeros along the real axis (at z = nπ)!")
````

### Rational with Pole: $f(z) = \frac{1}{z}$

For functions with poles at the origin, use an Annulus domain to avoid numerical issues:

````python
def reciprocal(z):
    return 1 / z

# Use annulus to avoid pole at origin
annulus_domain = cp.Annulus(inner_radius=0.2, outer_radius=2)
cmap = cp.Phase(phase_sectors=12, auto_scale_r=True)

fig = cp.pair_plot(annulus_domain, reciprocal, cmap=cmap, title=r"$f(z) = \frac{1}{z}$", resolution=600)
plt.tight_layout()
plt.show()

print("Notice how f(z) = 1/z reverses the phase direction!")
print("Also note the modulus inversion: |1/z| = 1/|z|")
````
