# Modulus Scaling for 3D and Riemann Sphere

**Prerequisites**: Notebooks #3, #6 | **Time**: ~25 min

```python
import numpy as np
import matplotlib.pyplot as plt
import complexplorer as cp

def f(z):
    return (z - 1) / (z**2 + z + 1)

domain = cp.Disk(radius=2)
cmap = cp.Phase(phase_sectors=6, auto_scale_r=True)
```

## Why Modulus Scaling?

**Problem**: |f(z)| ranges from 0 to ∞ (unbounded)

**Solution**: Map modulus to bounded range for 3D visualization

## All Scaling Modes (matplotlib 3D)

```python
modes = [
    ('none', 'Direct modulus'),
    ('constant', 'Flat (phase only)'),
    ('arctan', 'Smooth bounded'),
    ('logarithmic', 'Emphasize poles/zeros'),
    ('linear_clamp', 'Linear with limit'),
    ('adaptive', 'Percentile-based'),
]

fig = plt.figure(figsize=(15, 10))

for i, (mode, desc) in enumerate(modes):
    ax = fig.add_subplot(2, 3, i + 1, projection='3d')
    cp.plot_landscape(domain, f, cmap=cmap, modulus_mode=mode, ax=ax, z_max=5, resolution=100)
    ax.set_title(f"{desc}\n(mode='{mode}')", fontsize=10)
    ax.view_init(elev=25, azim=45)

plt.suptitle('Modulus Scaling Modes Comparison', fontsize=14)
plt.tight_layout()
plt.show()

print("\n✓ Each mode emphasizes different features!")
```

## Mode Descriptions

- **none**: Direct |f(z)| (can be very large)
- **constant**: Fixed height (phase only)
- **arctan**: Smooth, bounded (good default)
- **logarithmic**: Emphasize poles/zeros
- **linear_clamp**: Linear then clamp
- **adaptive**: Auto-adjust to data
- **power**, **sigmoid**, **hybrid**, **custom**: See docs

## Parameter Exploration: Arctan Mode

```python
params_list = [
    {'r_min': 0.5, 'r_max': 1.5},
    {'r_min': 0.2, 'r_max': 1.0},
    {'r_min': 0.8, 'r_max': 2.0},
    {'r_min': 0.5, 'r_max': 3.0},
]

fig, axes = plt.subplots(2, 2, figsize=(12, 12), subplot_kw={'projection': '3d'})
axes = axes.flatten()

for i, params in enumerate(params_list):
    cp.plot_landscape(
        domain, f, cmap=cmap,
        modulus_mode='arctan',
        modulus_params=params,
        ax=axes[i],
        resolution=100
    )
    axes[i].set_title(f"r_min={params['r_min']}, r_max={params['r_max']}", fontsize=11)
    axes[i].view_init(elev=25, azim=45)

plt.suptitle('Arctan Mode - Parameter Effects', fontsize=14)
plt.tight_layout()
plt.show()
```

## Custom Scaling Function

```python
# Custom: emphasize |f| ≈ 1
def custom_scale(moduli):
    return 1 + 0.3 * np.tanh(5 * (moduli - 1))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), subplot_kw={'projection': '3d'})

cp.plot_landscape(domain, f, cmap=cmap, modulus_mode='arctan', ax=ax1, resolution=100)
ax1.set_title('Standard arctan', fontsize=11)
ax1.view_init(elev=25, azim=45)

cp.plot_landscape(domain, f, cmap=cmap, modulus_mode='custom', 
                 modulus_params={'scaling_func': custom_scale}, ax=ax2, resolution=100)
ax2.set_title('Custom (emphasize |f|=1)', fontsize=11)
ax2.view_init(elev=25, azim=45)

plt.tight_layout()
plt.show()

print("\n✓ Custom scaling for specific analysis needs!")
```

## Use Case Guide

| Goal | Recommended Mode |
|------|------------------|
| General 3D | `arctan` (smooth, bounded) |
| Emphasize poles/zeros | `logarithmic` |
| Phase-only analysis | `constant` |
| STL export | See Notebook #8 |
